```python
# fib
import time
def fib(n):
  if n == 1 or n == 0:
    return 1
  else:
    return fib(n - 1) + fib(n - 2)
```

```python
start = time.time()
fib(1)
print(time.time() - start)
```

```python
start = time.time()
fib(12)
print(time.time() - start)
```

```python
start = time.time()
fib(36)
print(time.time() - start)
```

```python
start = time.time()
fib(40)
print(time.time() - start)
```

So this code is very inefficient

To improve this we have memoization

```python
# fib
import time
def fib(n, d):
  if n in d:
    return d[n]
  else:
    d[n] = fib(n - 1, d) + fib(n - 2, d)
    return d[n]
```

```python
start = time.time()
d = {0:1, 1:1}
fib(1, d)
print(time.time() - start)
```

```python
start = time.time()
d = {0:1, 1:1}
fib(12, d)
print(time.time() - start)
```

```python
start = time.time()
d = {0:1, 1:1}
fib(36, d)
print(time.time() - start)
```

```python
start = time.time()
d = {0:1, 1:1}
fib(40, d)
print(time.time() - start)
```

Now after memoization, compare the time taken to complete the task, it is efficient now
