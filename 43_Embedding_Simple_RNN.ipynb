```python
docs = ['go india',
        'india india',
        'hip hip hurray',
        'jeeteega bhai jeetega india jeetega',
        'bharat mata ki jai',
        'kohli kohli',
        'sachin sachin',
        'dhoni dhoni',
        'rohit rohit',
        'inquilab jindabad']
```

```python
import tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts(docs)
```

```python
len(tokenizer.word_counts)
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
```

```python
sequences = pad_sequences(sequences, padding = 'post')
sequences
```

```python
from tensorflow.keras.layers import Input, Embedding
from tensorflow.keras.models import Sequential

model = Sequential()
model.add(Input(shape=(5,))) # Tell the model the input shape first
model.add(Embedding(18, output_dim=2))

model.summary()
```

```python
model.compile('adam', 'accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
import tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN,Embedding,Flatten

(X_train,y_train),(X_test,y_test) = imdb.load_data(num_words=10000)
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50, dtype = 'int32')
X_test = pad_sequences(X_test,padding='post',maxlen=50, dtype = 'int32')
```

```python
X_train.shape
```

```python
from tensorflow.keras.layers import Input

model = Sequential()
model.add(Input(shape=(50,)))
model.add(Embedding(10000, 2))
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```
