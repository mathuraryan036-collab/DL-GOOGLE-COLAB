```python
import numpy as np
import pandas as pd
import time
```

```python
df = pd.read_csv('/content/Social_Network_Ads.csv')
```

```python
df = df[['Age', 'EstimatedSalary', 'Purchased']]
```

```python
X = df.iloc[:, 0:2]
Y = df.iloc[:, -1]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
```

```python
model.add(Dense(10, activation='relu', input_dim=2))
model.add(Dense(10, activation='relu'))
model.add(Dense(1,  activation='sigmoid'))
```

```python
# for checking which converges faster batch gd

model.compile(loss = 'binary_crossentropy', metrics = ['accuracy'])
history = model.fit(X_scaled, Y, epochs = 10, batch_size = 400, validation_split = 0.2)
```

```python
# for checking which converges faster stochastic gd

model.compile(loss = 'binary_crossentropy', metrics = ['accuracy'])
history = model.fit(X_scaled, Y, epochs = 10, batch_size = 1, validation_split = 0.2)
```
